```python
# -*- coding: utf-8 -*-

#%% Libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# %% Importing Data
flights_data = pd.read_csv('flights.csv')
flights_data.head(10)
weather_data_pd = pd.read_csv('weather.csv')
weather_data_np = weather_data_pd.to_numpy()

```

```python
# %% Pandas Data Filtering/Sorting Question Answering. Use flights_data

# Question 1: How many flights were there from JFK to SLC? Int q_1 

# Use Boolean filtering to select flights and shape[0] to count the matching rows.
q_1 = flights_data[
    (flights_data["origin"] == "JFK") & 
    (flights_data["dest"] == "SLC")
].shape[0]

print(q_1)
print(type(q_1))
```

```python
# Question 2: How many airlines fly to SLC? Should be int q_2 

# Use Boolean filtering to select flights to SLC and nunique() to count the distinct carriers.
q_2 = flights_data[
    flights_data["dest"] == "SLC"
]["carrier"].nunique()

print(q_2)
print(type(q_2))
```

```python
# Question 3: What is the average arrival delay for flights to RDU? float q_3

# Use Boolean filtering to select flights to RDU and mean() to calculate their average arrival delay.
q_3 = flights_data[
    flights_data["dest"] == "RDU"
]["arr_delay"].mean()

print(q_3)
print(type(q_3))
```

```python
# Question 4: What proportion of flights to SEA come from the two NYC airports (LGA and JFK)?  float q_4

# Use Boolean filtering to select flights to SEA and shape[0] to count the matching rows.
sea_sum = flights_data[
    flights_data["dest"] == "SEA"
].shape[0]

# Use Boolean filtering with isin() to select LGA/JFK-to-SEA flights and shape[0] to count the matching rows.
sea_nyc = flights_data[
    (flights_data["dest"] == "SEA") & 
    (flights_data["origin"].isin(["LGA","JFK"]))
].shape[0]

# Divide LGA/JFK-to-SEA flights by all SEA-bound flights to calculate their proportion
q_4 = sea_nyc / sea_sum


print(q_4)
print(type(q_4))

```

```python
# Question 5: Which date has the largest average departure delay? Pd slice with date and float
# please make date a column. Preferred format is 2013/1/1 (y/m/d) q_5 

# Use astype(str) and string concatenation to combine year, month, and day into a new date column.
flights_data["date"] = (
    flights_data["year"].astype(str) + "/" +
    flights_data["month"].astype(str) + "/" +
    flights_data["day"].astype(str)
)

# Use groupby() to group flights by date and mean() to calculate the average departure delay for each day. 
daily_delay = flights_data.groupby("date")["dep_delay"].mean() 

# Use idxmax() to find the date with the highest average delay (and max() to retrieve that delay value).
max_date = daily_delay.idxmax()
# max_delay = daily_delay.max()

# Use loc[] to select the highest delay date while preserving the result as a one-element series.
q_5 = daily_delay.loc[[max_date]]

print(q_5)

```

```python
# Question 6: Which date has the largest average arrival delay? pd slice with date and float q_6

# Use groupby() to group flights by date and mean() to calculate the average arrival delay for each day.
largest_delay = flights_data.groupby("date")["arr_delay"].mean()

# Use idxmax() to find the date with the highest average arrival delay ( and max() to retrieve that delay value).
max_date = largest_delay.idxmax()
# max_delay = largest_delay.max()

# Use loc[] to select the highest delay date while preserving the result as a one-element series.
q_6 = largest_delay.loc[[max_date]]

print(q_6)

```

```python
# Question 7: Which flight departing LGA or JFK in 2013 flew the fastest? pd slice with tailnumber and speed
# speed = distance/airtime q_7

# Use Boolean filtering with isin() to select 2013 flights departing from either LGA or JFK.
filtered = flights_data[
    (flights_data["year"] == 2013) &
    (flights_data["origin"].isin(["LGA","JFK"]))
]

# Create an independent copy and add a speed column by dividing distance by air_time.
filtered = filtered.copy()
filtered["speed"] = filtered["distance"] / filtered["air_time"]

# Use idxmax() to find the fastest flight and loc[] to return its tail number and speed as a DataFrame.
q_7 = filtered.loc[
    [filtered["speed"].idxmax()],
    ["tailnum", "speed"]
]

print(q_7)

```

```python
# Question 8: Replace all nans in the weather pd dataframe with 0s. Pd with no nans q_8

# Use fillna(0) to replace all missing values in the weather DataFrame with zero.
q_8 = weather_data_pd.fillna(0)

print(q_8)
```

```python
# %% Numpy Data Filtering/Sorting Question Answering
# Use weather_data_np
# Question 9: How many observations were made in February? Int q_9 

# # Use to_numpy() to convert the Pandas weather DataFrame into a NumPy array.
weather_data_np = weather_data_pd.to_numpy()

# Create a Boolean mask and use np.sum() to count rows where the month equals 2.
q_9 = np.sum(weather_data_np[:,3] == 2)

print(q_9)
print(type(q_9))

```

```python
# Question 10: What was the mean for humidity in February? Float q_10

# Use Boolean indexing to select all weather rows where the fourth column (month) equals 2.
february = weather_data_np[weather_data_np[:,3] == 2]

# Use NumPy column slicing and mean() to calculate the average humidity for February.
q_10 = february[:, 8].mean()

print(q_10)
print(type(q_10))
```

```python
# Question 11: What was the std for humidity in February? Float q_11

february = weather_data_np[weather_data_np[:, 3] == 2]

# Use NumPy column slicing and std() to calculate the standard deviation of February humidity.
q_11 = february[:, 8].std() 

print(q_11)
print(type(q_11))
```
