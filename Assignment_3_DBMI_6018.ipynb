```python
# Problem 1: Lists, Sets and Coercion

one_a = list(range(10))
print(one_a)

one_b = one_a.copy()
one_b[5] += 3
print(one_b)

one_c = [float(number) for number in one_b]
print(one_c)

one_d = set(one_c)
print(one_d)

one_e = one_d.copy()
one_e.add(10)
print(one_e)

one_f = one_e.copy()
one_f.pop()
print(one_f)

one_g = len(one_f)
print(one_g)

one_h = len(one_f) == len(one_c)
print(one_h)

one_i = list(one_f) + one_a
print(one_i)

one_j = set(one_i)
print(one_j)

one_k = len(one_j)
print(one_k)
```
