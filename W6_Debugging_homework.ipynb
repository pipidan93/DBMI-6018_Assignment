```python
# Problem 1.a: Use print() to locate the error

def wrong_add_function(arg1, arg2):
    
    original_arg1 = arg1.copy()

    arg1_index = 0
    while arg1_index < len(arg1):
        arg_2_sum = 0

        for arg2_elements in arg2:
            arg_2_sum = sum([arg1[arg1_index] + i for i in arg2])

        arg1[arg1_index] = arg_2_sum
        arg1_index += 1
        
    expected = [original_arg1[index] + arg2[index] for index in range(len(original_arg1))]
    if arg1 != expected:
        print("We are making an error in the loop.")
        print(f"The correct answer is supposed to be: " f"{expected}")

    return arg1


arg1 = [4, 6, 8, 10]
arg2 = [2, 2, 2, 2]

problem1_a = wrong_add_function(arg1.copy(), arg2)

print(problem1_a)
```

```python
# Problem 2.b

def exception_add_function(arg1, arg2):
    try:
        expected_type = type(arg1[0])
        error_messages = []

        for argument_number, argument in enumerate((arg1, arg2), start=1):
            for element_index, value in enumerate(argument):
                if type(value) != expected_type:
                    error_messages.append(
                        f"Your input argument {argument_number} "
                        f"at element {element_index} "
                        f"is not of the expected type. "
                        f"Please change this and rerun."
                    )

        if error_messages:
            raise TypeError("\n".join(error_messages))

        if (sum([type(i) == int for i in arg1]) == len(arg1) and \
            sum([type(i) == int for i in arg2]) == len(arg2)):
            arg1_index = 0

            while arg1_index < len(arg1):
                arg_2_sum = (arg1[arg1_index] + arg2[arg1_index])

                arg1[arg1_index] = arg_2_sum
                arg1_index += 1

            return arg1

        elif (sum([type(i) == str for i in arg1]) == len(arg1) and \
              sum([type(i) == str for i in arg2]) == len(arg2)):
            arg1_index = 0

            while arg1_index < len(arg1):
                arg_2_sum = ""

                for arg2_elements in arg2:
                    arg_2_sum += arg2_elements

                arg1[arg1_index] = (arg1[arg1_index] + str(arg_2_sum))    

                arg1_index += 1

            return arg1

    except TypeError as error:
        return str(error)


arg_str_1 = [4, 6, "8", "10"]
arg_str_2 = ["2", "2", 2, 2]

problem2_b = exception_add_function(arg_str_1, arg_str_2)

print(problem2_b)
```

```python
# Problem 2.c

def correction_add_function(arg1, arg2):
    try:
        if sum([type(i) == int for i in arg1]) == len(arg1) and \
           sum([type(i) == int for i in arg2]) == len(arg2):

            arg1_index = 0

            while arg1_index < len(arg1):
                arg_2_sum = (arg1[arg1_index] + arg2[arg1_index])

                arg1[arg1_index] = arg_2_sum
                arg1_index += 1

            return arg1

        elif sum([type(i) == str for i in arg1]) == len(arg1) and \
             sum([type(i) == str for i in arg2]) == len(arg2):

            arg1_index = 0

            while arg1_index < len(arg1):
                arg_2_sum = ""

                for arg2_elements in arg2:
                    arg_2_sum += arg2_elements

                arg1[arg1_index] = (
                    arg1[arg1_index] + str(arg_2_sum))

                arg1_index += 1

            return arg1

        else:
            raise TypeError

    except TypeError:
        corrected_arg1 = [str(value) for value in arg1]
        corrected_arg2 = [str(value) for value in arg2]

        return correction_add_function(corrected_arg1, corrected_arg2)


arg_str_1 = ["4", "6", "8", "10"]
arg_str_2 = ["2", "2", 2, 2]

# arg_str_1 = [4, 6, 8, 10]
# arg_str_2 = [2, 2, 2, 2]

problem2_c = correction_add_function(arg_str_1, arg_str_2)

print(problem2_c)
```
